```python
import matplotlib.pyplot as plt
import numpy as np

# Sample dataset
months = ['Jan', 'Feb', 'Mar', 'Apr', 'May', 'Jun']
sales = [12000, 15000, 13500, 18000, 22000, 25000]

products = ['Laptop', 'Mobile', 'Tablet', 'Watch']
product_sales = [45000, 60000, 30000, 20000]

advertising = [10, 20, 30, 40, 50, 60, 70]
revenue = [15, 25, 35, 45, 50, 65, 75]

employee_salaries = [
    25000, 30000, 35000, 40000, 45000,
    50000, 55000, 60000, 65000, 70000
]

# Create multiple subplots
fig, axes = plt.subplots(2, 2, figsize=(14, 10))

# 1. Line Plot
axes[0, 0].plot(
    months, sales,
    marker='o',
    linestyle='--',
    label='Monthly Sales'
)

axes[0, 0].set_title('Monthly Sales Trend')
axes[0, 0].set_xlabel('Months')
axes[0, 0].set_ylabel('Sales (₹)')
axes[0, 0].legend()
axes[0, 0].grid(True)

# 2. Bar Chart
axes[0, 1].bar(
    products,
    product_sales,
    label='Product Sales'
)

axes[0, 1].set_title('Product-wise Sales')
axes[0, 1].set_xlabel('Products')
axes[0, 1].set_ylabel('Sales (₹)')
axes[0, 1].legend()

# 3. Scatter Plot
axes[1, 0].scatter(
    advertising,
    revenue,
    marker='o'
)

axes[1, 0].set_title('Advertising vs Revenue')
axes[1, 0].set_xlabel('Advertising Cost')
axes[1, 0].set_ylabel('Revenue')
axes[1, 0].grid(True)

# 4. Histogram
axes[1, 1].hist(
    employee_salaries,
    bins=5,
    edgecolor='black'
)

axes[1, 1].set_title('Employee Salary Distribution')
axes[1, 1].set_xlabel('Salary')
axes[1, 1].set_ylabel('Frequency')

# Layout adjustment
plt.tight_layout()

# Save visualization
plt.savefig('matplotlib_analysis.png', dpi=300)

# Display all plots
plt.show()
```
