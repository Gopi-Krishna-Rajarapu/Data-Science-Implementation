```python
import numpy as np

print("========== NUMPY OPERATIONS FOR DATA SCIENCE ==========")

# 1. Creating Arrays
arr = np.array([10, 20, 30, 40, 50])

print("\n1. Original Array:")
print(arr)

# 2. Array Properties
print("\n2. Array Properties:")
print("Dimensions:", arr.ndim)
print("Shape:", arr.shape)
print("Size:", arr.size)
print("Data Type:", arr.dtype)

# 3. Mathematical Operations
print("\n3. Mathematical Operations:")
print("Addition:", arr + 5)
print("Subtraction:", arr - 5)
print("Multiplication:", arr * 2)
print("Division:", arr / 2)
print("Square:", arr ** 2)

# 4. Statistical Operations
print("\n4. Statistical Operations:")
print("Sum:", np.sum(arr))
print("Mean:", np.mean(arr))
print("Median:", np.median(arr))
print("Minimum:", np.min(arr))
print("Maximum:", np.max(arr))
print("Standard Deviation:", round(np.std(arr), 2))
print("Variance:", np.var(arr))
```
