```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('ecommerce_sales_dataset.xlsx')
df.columns
```

```python
# Countplot – Gender-wise category counts
sns.countplot(data=df, x="Category", hue="Gender")
plt.show()
```

```python
sns.countplot(data =df df.Gender)
```

```python
# Violinplot – Product price distribution
sns.violinplot(data=df, x="Category", y="Price")
plt.show()
```

```python
# Boxplot – Customer ratings
sns.boxplot(data=df, x="Category", y="Rating")
plt.show()
```

```python
# Lineplot – Average rating by product
sns.lineplot(data=df, x="Product", y="Rating", estimator="mean")
plt.xticks(rotation=45)
plt.show()
```

```python
# Stripplot – Individual ratings by category
sns.stripplot(data=df, x="Category", y="Rating")
plt.show()
```

```python
# Swarmplot - Ratings by gender
sns.swarmplot(data=df, x="Gender", y="Rating")
plt.show()
```

```python
# Pointplot – Average product prices across cities
sns.pointplot(data=df, x="City", y="Price", estimator="mean")
plt.xticks(rotation=45)
plt.show()
```
