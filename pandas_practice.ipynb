```python
import pandas as pd
df = pd.read_excel("Movies_Dataset.xlsx") 
df
```

```python
df.columns
```

```python
# 1.Select all movies in the "Action" genre. 
df[df.Genre == "Action"]['Movie Name']
```

```python
df.dtypes
```

```python
# 2. Select movies released after 2000. 
df.loc[df['Release_Year'] > 2000, 'Movie Name']
```

```python

# 3. Select movies with an IMDb rating greater than 8.5.
df[df.IMDb_Rating > 8.5]
```

```python
# 4. Select only "Movie Name" and "Genre" columns. 
df[["Movie Name","Genre"]]
```

```python
# 5. Select movies released between 1990 and 2010.
df.loc[(df['Release_Year'] >= 1990) & (df['Release_Year'] <= 2010), 'Movie Name']    
# df[(df.Release_Year >= 1990) |  (df.Release_Year <= 2010)] 
```

```python
# 6.Count the number of movies in each genre. 
print(df.Genre.value_counts())
```

```python
# 7. Find the highest IMDb-rated movie. 
# df.loc[df['IMDb_Rating'].idxmax(), ['Movie Name', 'IMDb_Rating']]
df.loc[df.IMDb_Rating.idxmax(), ['Movie Name','IMDb_Rating']] # (or) df.loc[df[IMDb_Rating]idxmax(), ['Movie Name','IMDb_Rating']]
```

```python
# 8. Find the average IMDb rating of all movies. 
df['IMDb_Rating'].mean() # df.IMDb_Rating.mean()
```

```python
# 9. Find the total box office collection of all movies. 
print(df.Revenue.sum())
```

```python
# 10. Select movies with box office earnings greater than 500 million. 
df[df['Revenue'] > 500]['Movie Name']
```
