```python
import pandas as pd

df =  pd.read_excel('Movies_Dataset.xlsx')
df
```

```python
# 11. Rank movies based on IMDb rating. 

df['Rank'] = df['IMDb_Rating'].rank( ascending = False, method = 'dense')
df = df.sort_values('Rank')
df
```

```python
# 12. Sort movies by release year in descending order.
df = df.sort_values('Release_Year')
df
```

```python
# 13. Find all duplicate movie names (if any).
# df.value_counts(df.duplicated('Movie Name'))
df[df.duplicated('Movie Name')]
```

```python
# 14. Remove duplicate movie names. 

[df.duplicated(['Movie Name'])]
df=df.drop_duplicates(subset ='Movie Name')
print(df.value_counts(df.duplicated('Movie Name')))
print(df[df.duplicated(subset='Movie Name')])
```

```python
# 15. Replace "Sci-Fi" genre with "Science Fiction".

print(df['Genre'].head(20).to_list())

# df = df.replace({'Genre' : { 'Sci-Fi' : 'Science Fiction'}})
# df.loc[df['Movie Name'] == 'Science Friction']  
# df = df.rename(column ={'Genre' : 'Jonor'}})
df['Genre'] = df['Genre'].replace({'Sci-Fi' : 'Science Fiction'})
# print(df['Genre'].str.contains('Science Fiction', case = False, na= False))
print(df['Genre'].head(20).to_list())
```
