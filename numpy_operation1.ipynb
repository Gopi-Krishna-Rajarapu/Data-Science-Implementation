```python

import pandas as pd
import numpy as np
df = pd.read_csv("Indian_Cricket_Team_Dataset.csv")
df
```

```python
# Query 1: Convert 'Age' column to a NumPy array. 
age_array = np.array(df['Age'])
age_array
```

```python
# Query 2: Find the average age of players. 
avg_age = np.mean(age_array)
print(avg_age)
```

```python
# Query 3: Find the oldest and youngest player age. 
young_p = np.min(age_array)
old_p   = np.max(age_array)
print(young_p )
print(old_p)
```

```python
# Query 4: Count the total number of players above 30 years old. 
count_ab_30 = np.sum(age_array > 30) 
print(count_ab_30)
```

```python
# Query 5: Find the median batting average. 
bat_avg = np.array(df['Batting Average'])
avg_bat = np.median(bat_avg)
print(avg_bat)
```

```python
# Query 6: Find the standard deviation of strike rates. 
strike_rate = np.array(df['Strike Rate'])
std_strike = np.std(strike_rate)
print(std_strike)
```

```python
# Query 7: Find the 90th percentile of batting average.
print(np.percentile(bat_avg, 100))
print(np.max(bat_avg))
```

```python
# Query 8: Find the number of players with strike rate above 150. 
count_ab_150 = np.sum(strike_rate > 150)
print(count_ab_150)
```

```python
# Query 9: Reshape the first 20 batting averages into a 4x5 matrix. 
bat_avg = np.array(df['Batting Average'].head(20))
avg_reshape= bat_avg.reshape(4,5)
avg_reshape
```

```python
# Query 10: Compute row-wise sum of reshaped matrix. 
bat_avg = np.array(df['Batting Average'].head(20))
bat_matrix = bat_avg.reshape(4, 5)

row_sum = np.sum(bat_matrix, axis=1)
col_sum = np.sum(bat_matrix, axis=0)

print(row_sum)
print(col_sum)
```
