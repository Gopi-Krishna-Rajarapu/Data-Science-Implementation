```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_excel('ecommerce_sales_dataset.xlsx')
df
```

```python
# 1. Create a histogram of customer ages using histplot().
# To find shape of data
sns.histplot(data = df, x = "Age", bins = 10, kde = True)
```

```python
# 3. Display the average product price for each category using barplot()
# To find the frequency of categorical data
sns.countplot(data = df, x = "Category")
plt.title("Number of Orders by Category")
plt.xlabel("Category")
plt.ylabel("Number of Orders")
plt.show()
```

```python
# 3. Display the average product price for each category using barplot()
sns.barplot(data = df , x = "Category",y = "Price")
```

```python
sns.boxplot(data = df.Price)
plt.xlabel("Outliers in Prices ")
plt.ylabel("Prices ")
```

```python
# 5. Use scatterplot() to compare customer age with product price.
sns.scatterplot(data = df, y = "Price", x = 'Age')
```

```python
# 6. Create a Matplotlib pie chart showing the share of orders by category.
data = pd.DataFrame(df)
category_count = df["Category"].value_counts()
print(category_count)
plt.pie(category_count, labels = category_count.index)
```
